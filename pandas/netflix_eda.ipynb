```python
import pandas as pd
import matplotlib.pyplot as plt

netflix_df = pd.read_csv("netflix_titles.csv")

type_counts = netflix_df["type"].value_counts()

type_counts.plot(
    kind="bar",
    color=["steelblue", "orange"]
)

plt.title("Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
```

Movies greatly outnumber TV Shows on Netflix.

```python
import pandas as pd
import matplotlib.pyplot as plt

netflix_df = pd.read_csv("netflix_titles.csv")

type_counts = netflix_df["type"].value_counts()

type_counts.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Netflix Content Types")

plt.show()
```

Movies make up the majority of Netflix content.

```python
import pandas as pd
import matplotlib.pyplot as plt

netflix_df = pd.read_csv("netflix_titles.csv")

titles_per_year = (
    netflix_df["release_year"]
    .value_counts()
    .sort_index()
)

titles_per_year.plot(kind="line")

plt.title("Netflix Titles Over Time")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

The titles per release year rose sharply until about 2018, then dropped, and the final year is incomplete.

```python
import pandas as pd
import matplotlib.pyplot as plt

netflix_df = pd.read_csv("netflix_titles.csv")

rating_counts = netflix_df["rating"].value_counts()

rating_counts.plot(
    kind="bar",
    color="purple"
)

plt.title("Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")

plt.show()
```

TV-MA and TV-14 appear among the most common ratings.

```python
import pandas as pd
import matplotlib.pyplot as plt

netflix_df = pd.read_csv("netflix_titles.csv")

plt.hist(
    netflix_df["release_year"],
    bins=20
)

plt.title("Distribution of Release Years")
plt.xlabel("Release Year")
plt.ylabel("Frequency")

plt.show()
```

Most titles were released in more recent years.

```python
import pandas as pd
import matplotlib.pyplot as plt

netflix_df = pd.read_csv("netflix_titles.csv")

country_counts = (
    netflix_df["country"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

country_counts.plot(
    kind="bar",
    color="green"
)

plt.title("Top 10 Countries by Number of Titles")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.show()
```

The United States contributes the highest number of titles.

########## KEY FINDINGS ##########

1. Movies outnumber TV Shows.

2. Netflix content increased rapidly in recent years.

3. TV-MA is one of the most common ratings.

4. Most titles were released after 2000.

5. The United States produces the most Netflix content.
