```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("netflix_titles.csv")

counts = df["type"].value_counts()

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(counts.index, counts.to_numpy())
ax.set_title("Titles by type")
ax.set_xlabel("Type")
ax.set_ylabel("Number of titles")
fig.savefig("titles_by_type.png", dpi=150)

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
df = pd.read_csv("netflix_titles.csv")

counts = df["type"].value_counts()
print(counts)
print((df["type"].value_counts(normalize=True) * 100).round(1))

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="type", order=counts.index)
plt.title("Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Number of titles")
plt.show()
```
